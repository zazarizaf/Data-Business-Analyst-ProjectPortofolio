```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
import pyarrow as pa
```

```python
glob.glob('D:/MBB 28/Data Analysis Portofolio Project/dunhumby/retail/datasets/datasets_retail/transactions_*.csv')
```

```python
all_dfs=[]

for df in glob.glob('D:/MBB 28/Data Analysis Portofolio Project/dunhumby/retail/datasets/datasets_retail/transactions_*.csv'):
    print(f'Loading {df}')
    new_df=pd.read_csv(df)

    all_dfs.append(new_df)
```

```python
df=pd.concat(all_dfs)
df.to_parquet('dunnhumby_clean.parquet', compression='snappy')
```

```python
df=pd.read_parquet('dunnhumby_clean.parquet')
display(df.info(), df.head(5), df.columns)
```

```python
print(df['SPEND'].notna().sum(), df['SFNND'].notna().sum())
print((df['SPEND'].notna() & df['SFNND'].notna()).sum())
```

```python
df['SPEND']=df['SPEND'].fillna(df['SFNND'])
df['BASKET_TYPE']=df['BASKET_TYPE'].fillna(df['BASKET_TYFN'])

df = df.drop(columns=['SFNND', 'BASKET_TYFN'])
```

```python
df
```

```python
df['SHOP_WEEK']=df['SHOP_WEEK'].astype(str)
df['SHOP_DATE']=pd.to_datetime(df['SHOP_DATE'], format='%Y%m%d')
```

```python
df[df.duplicated()]
```

```python
df=df.drop_duplicates()
```

```python
df['BASKET_ID']=df['BASKET_ID'].astype('int64')
```

```python
df
```

```python
weekdays={
    1:'Sunday',
    2:'Monday',
    3:'Tuesday',
    4:'Wednesday',
    5:'Thursday',
    6:'Friday',
    7:'Saturday'
}

df['SHOP_WEEKDAY']=df['SHOP_WEEKDAY'].map(weekdays).astype('str')
df
```

```python
def group_time(x):
    if 1 <= x <= 6:
        return 'Morning'
    elif 7 <= x <= 12:
        return 'Afternoon'
    elif 13 <= x <= 18:
        return 'Evening'
    else:
        return 'Night'

df['SHOP_HOUR_GROUP']=df['SHOP_HOUR'].apply(group_time)
df
```

```python
df.to_csv('D:/MBB 28/Data Analysis Portofolio Project/dunhumby/retail/datasets/datasets_retail/cleaned_dunnhumby.csv', 
          index=False,
          sep=';',
          decimal=',')
```
